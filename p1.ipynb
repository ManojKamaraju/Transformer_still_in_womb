```python
def isDistinct(l : list) -> bool: # Return True if All are same else False
    count = 0
    for i in range(len(l)):
        for j in range(len(l)):
            if(l[i] == l[j]):
                count += 1
    return True if(count == (len(l) * len(l))) else False

    
print(isDistinct([4,4,2]))
```

```python
def isDistinct(l : list) -> bool: # Return True if All are same else False
    for i in range(len(l)):
        for j in range(1,len(l)):
            if(l[i] != l[j]):
                return False
    return True
```

```python
print(isDistinct([8, 8, 7]))
print(isDistinct([8,8,8]))
```

```python
if not isDistinct(a):
    b = list(set(a))

    if len(b) < 2:
        print("No second-largest distinct element")
    else:
        for i in range(len(b)):
            count = 0

            for j in range(len(b)):
                if b[i] > b[j]:
                    count += 1

            if count == len(b) - 2:
                print("Second Largest Distinct element is:", b[i])
```
