**Byte Pair Encoding 1**

```python
class BPE1:
    def __init__(self , sample):
        self.vocab = {}
        self.pairs = {}
        self.adjacent_pairs = []
        self.max_rep = 0
        self.token = 0
        self.sample = sample

    def count_pairs(self, sample = None): 
        if sample is None:
            sample = self.sample
        sample = list(sample) 
        self.adjacent_pairs  = []
     
        num = 0
        for i in range(len(sample) -  1):
            self.adjacent_pairs.append((sample[i] , sample[i+1]))
        self.pairs = {}
        for i in range(len(self.adjacent_pairs ) ):            
            self.pairs[self.adjacent_pairs [i]] = self.adjacent_pairs.count(self.adjacent_pairs [i])
        return self.pairs


    def find_best_pair(self):
        self.max_rep = max(self.pairs , key=self.pairs.get)
        return self.max_rep
        

    
    def merge_pairs(self , tokens = None , pair = None ):
        merge = []
        i = 0
        if tokens == None and pair == None:
            tokens = self.sample
            pair = self.max_rep
        

        while i < len(tokens):
            if i < len(tokens) -1 and (tokens[i] , tokens[i+1]) == pair:
                merge.append(tokens[i] + tokens[i + 1])
                i += 2

            else:
                merge.append(tokens[i])
                i += 1
        return merge



    def train(self , it):
        final = []
        pair_best = []
        tokens = list(self.sample)
        for i in range(it):
            pair = self.count_pairs(tokens)

            if not pair:
                break
            best_pair = self.find_best_pair()
            tokens = self.merge_pairs(tokens , best_pair)
            final.append(tokens)
            pair_best.append(best_pair)

        return final ,  pair_best
        

        

sample = "Banana "
obj = BPE1(sample)
print(obj.count_pairs())
print(obj.find_best_pair())
print(obj.merge_pairs())

print(obj.train(6))
```
