```python
def Encoder(s : str , vocab : dict) -> list:
    tok = []
    for alphabet in s:
        if alphabet not in vocab:
            tok.append(vocab["<UNK>"])
        else:
            tok.append(vocab[alphabet])
    return tok

print(Encoder(actualtext[:100] , vocab)) # output [84, 104, 101, 32, 80, 114, 111, 106, 101, 99, 116, 32, 71, 117, 116, 101, 110, 98, 101, 114, 103, 32, 101, 66, 111, 111, 107, 32, 111, 102, 32, 65, 110, 32, 69, 110, 103, 108, 105, 115, 104, 32, 71, 114, 97, 109, 109, 97, 114, 10, 84, 104, 105, 115, 32, 101, 66, 111, 111, 107, 32, 105, 115, 32, 102, 111, 114, 32, 116, 104, 101, 32, 117, 115, 101, 32, 111, 102, 32, 97, 110, 121, 111, 110, 101, 32, 97, 110, 121, 119, 104, 101, 114, 101, 32, 105, 110, 32, 116, 104]
```

```python
def Decoder(tok : list , vocab : dict) -> str:

    ans = ""
    vocab = {v : k for k , v in vocab.items()}
    for alphabet in tok:
        if alphabet not in vocab:
            ans += vocab[-1]
        else:
            ans += vocab[alphabet]
    return ans


print(Decoder([1, 2, 3, 4, 5, 6, 7, 8, 3, 9, 10, 4, 11, 12, 10, 3, 13, 14, 3, 6, 15, 4, 3, 16, 7, 7, 17, 4, 7, 18, 4, 19, 13, 4, 20, 13, 15, 21, 22, 23, 2, 4, 11, 6, 24, 25, 25, 24, 6, 26, 1, 2, 22, 23, 4, 3, 16, 7, 7, 17, 4, 22, 23, 4, 18, 7, 6, 4, 10, 2, 3, 4, 12, 23, 3, 4, 7, 18, 4, 24, 13, 27, 7, 13, 3, 4, 24, 13, 27, 28, 2, 3, 6, 3, 4, 22, 13, 4, 10, 2] , vocab))
```
