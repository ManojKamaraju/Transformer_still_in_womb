**Making text from pdf**

```python
from pypdf import PdfReader
```

```python
reader = PdfReader("english_tb.pdf")

textbook = []


for page in reader.pages:
    text = page.extract_text()

    if text:
        textbook.append(text)

    
print(textbook)   
```

```python
print(len(textbook))
print("----------")
print(reader.pages)
```

```python
#Making text files into system
with open("english_textbook.txt" , "w") as f:
    for text in textbook:
        f.write(text)

with open("english_textbook.txt" , "r") as g:
    actualtext = g.read()
print(actualtext)
```
