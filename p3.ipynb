```python
def IO(l : list) -> dict:

    if len(l) < 1:
        raise ValueError("List is empty")
    minimum = l[0]
    maximum = l[0]
    for i in range(len(l)):
            if(minimum > l[i]):
                minimum  =  l[i]
                
            if(maximum < l[i]):
                maximum  =  l[i]
    
    return {"count" : len(l) , "Maximum" : maximum , "Minimum" : minimum , "Average" : sum(l) / len(l) }

IO(l)
```
