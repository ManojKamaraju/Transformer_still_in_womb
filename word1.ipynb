```python
class BasicWordTokenizer:
    def __init__(self ):
        self.vocab = {"<UNK>" : 0 , " " : 1}

    #Make Vocab
    def MakeVocab(self ,file) -> dict :
        num = 1
        with open(file , "r") as f:
            text = f.read()
        for t in text.split():
            if t not in vocab:
                num += 1
                vocab[t] = num
        return vocab , text

    #Make Encoder
    def MakeEncoder(self,string : str) -> list:
        l = []
        for s in string.split():
            if s in self.vocab:
                l.append(self.vocab[s])
                l.append(self.vocab[" "])
            else:
                l.append(self.vocab["<UNK>"])
                l.append(self.vocab[" "])
        return l

    # Make Decoder
    def MakeDecoder(self , seq : list) -> str:
        s = ""
        rev_vocab = {v : k for k,v in self.vocab.items()}
        for se in seq.split():
            if se not in rev_vocab:
                s += rev_vocab[0]
                s += rev_vocab[1]
            else:
                s += rev_vocab[se]
                s += rev_vocab[se]
        return s

        
            
        
        
        
obj = BasicWordTokenizer()      
```

```python
vocab , text = obj.MakeVocab("english_textbook.txt")
print(text[:21]) #The Project Gutenberg
```
