**Problem decomposition**<br>
Programming problem: Take “analyze a text file” and split it into small functions with clear inputs and outputs; implement the smallest working version.
<br>
Problem statement
Write a Python program that reads a text file and analyzes its contents by calculating:
1. Character count: Total number of characters, including spaces and newline characters.
2. Word count: Number of whitespace-separated words.
3. Line count: Number of lines in the text.
Your program must be divided into small functions, each with a clear responsibility, input, and output.<br>
Function	Input	Expected output
read_file(path)	File path (str)	File contents (str)
count_characters(text)	Text (str)	Character count (int)
count_words(text)	Text (str)	Word count (int)
count_lines(text)	Text (str)	Line count (int)
main()	No arguments required	Displays the analysis report

Line count: Number of lines in the text.

```python


def countLine(s : str) :
    #opening file
    
    with open(s, 'r+' ) as f:
        return len(f.readlines())
    

countLine('sample.txt' )
#got output as 39

 
```

Count words

```python


def countWord(s : str) :
    #opening file
 
    with open(s, 'r+' ) as f1:
        return len(f1.read().split())
 

countWord('sample.txt' )
#got output as 645
```
