```python
word = "Hello World"
word = word.lower()

vocab = {"<UNK>" : 0}
num = 1
for w in word:
    vocab[w] = num
    num += 1
print(vocab)
```

```python
word = "Hello World"
def MakeVocab(s : str) -> dict:
    vocab = {"<UNK>" : 0}
    num = 1
    s = s.lower()
    for w in s : 
        if w not in vocab:
            vocab[w] = num
            num += 1
    return vocab
vocab = MakeVocab(word)
print(vocab)
        
    
```

```python
def Encode(s : str , vocab:dict) -> list:
    s = s.lower()
    ans = []
    for w in s:
        ans.append(vocab[w])
    return ans

token = Encode(word,vocab)
print(token)
```

```python
def Decode(l : list  , vocab : dict) -> str:
    ans = ""
    vocab = {v : k for k , v in vocab.items()}
    
    for i in l:
        if i in vocab:
            ans += vocab[i]
        else:
            ans += vocab[0]
    return ans
print(Decode(token , vocab))
```
